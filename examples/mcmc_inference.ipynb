# U-MIMIC MCMC Example

This notebook shows a basic Bayesian workflow using the inference extras.

```python
from umimic.pipeline.config import ExperimentConfig
from umimic.pipeline.experiment import Experiment
```

```python
config = ExperimentConfig(
    name="mcmc_example",
    dosing={"concentrations": [0, 0.3, 1.0, 3.0]},
    simulation={"method": "ode", "t_max": 96.0},
    inference={"mode": "mcmc", "n_samples": 500, "n_burnin": 200},
)
exp = Experiment(config)
dataset = exp.generate_synthetic()
posterior = exp.fit(dataset)
posterior
```
