# U-MIMIC Quickstart (API)

This notebook demonstrates a minimal in-memory simulation and fit workflow.

```python
from umimic.pipeline.config import ExperimentConfig
from umimic.pipeline.experiment import Experiment
```

```python
config = ExperimentConfig(
    name="quickstart",
    dosing={"concentrations": [0, 0.1, 1.0, 10.0]},
    simulation={"method": "ode", "t_max": 72.0},
    inference={"mode": "mle", "n_restarts": 3},
)
exp = Experiment(config)
dataset = exp.generate_synthetic()
result = exp.fit(dataset)
result.mle.parameters
```
